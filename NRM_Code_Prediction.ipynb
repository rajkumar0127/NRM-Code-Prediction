# Removing Punctuations

```python
#function to remove punctuations from the sentences
def remove_punctuation(text):
  translator=str.maketrans("","",string.punctuation)
  return text.translate(translator)
```

```python
#get values without punctuation
x=[]
for text in x_data:
  without_punctuation=remove_punctuation(str(text)
  x.append(without_punctuation.lower())
```

```python
len(x)
```

# Removing Numbers from Dataset

```python
#function to remove numbers
def remove_numbers(text):
    text_without_numbers = re.sub(r'\d+', '', text)
    return text_without_numbers
```

```python
#get values without numbers
x1=[]
for text in x:
  without_num=remove_numbers(text)
  x1.append(without_num)
```

```python
len(x1)
```

# Lemmatization and Removing StopWords



```python
stop_words=set(stopwords.words("english"))
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
def remove_stopwords(text):
    # Tokenize the text into individual words
    tokens = word_tokenize(text)

    # Remove stopwords from the tokens
    filtered_tokens = [lemmatizer.lemmatize(word) for word in tokens if word.lower() not in stop_words]

    # Join the filtered tokens back into a single text string
    text_without_stopwords = ' '.join(filtered_tokens)

    return text_without_stopwords
```

```python
x_token=[]
for text in x1:
  without_stopwords=remove_stopwords(text)
  x_token.append(without_stopwords)
```

# Vectorization

```python
tfid=TfidfVectorizer()
```

```python
#transforming data into vectorized form
x_tfid=tfid.fit_transform(x_token)
```

# Train Test Split

```python
#split data into train & test
x_train,x_test,y_train,y_test=tts(x_tfid,y_data,test_size=.1,random_state=1)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```
